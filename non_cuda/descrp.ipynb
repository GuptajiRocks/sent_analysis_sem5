```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer

# ---------------------------------------------
# 1️⃣ Load All Datasets
# ---------------------------------------------
# Adjust the file paths if needed
files = {
    "airline": "airline.tsv",
    "eco_news": "eco_news.tsv",
    "glo_warm": "glo_warm.tsv",
    "text_emo": "text_emo.tsv"
}

dataframes = []
for name, path in files.items():
    try:
        df_temp = pd.read_csv(path, sep='\t', encoding="latin1")
        df_temp['source'] = name
        dataframes.append(df_temp)
        print(f"✅ Loaded {name} — {df_temp.shape}")
    except Exception as e:
        print(f"❌ Error loading {name}: {e}")

# Merge all datasets
df = pd.concat(dataframes, ignore_index=True)
print("\n🔹 Total merged shape:", df.shape)
```
